### 2. Bubble Sort

Implements Bubble Sort by repeatedly comparing adjacent elements and swapping them when they are out of order.

- **Approach:** Comparison-based, in-place sorting.
- **Time Complexity:** \(O(n^2)\)
- **Space Complexity:** \(O(1)\)
- **Order:** Ascending.

```python
def BubbleSort(list):
    listSize = len(list)
    for i in range(0, listSize - 1):
        for j in range(0, listSize - 1):
            if list[j] > list[j + 1]:
                list[j], list[j + 1] = list[j + 1], list[j]
        listSize = listSize - 1
    return list
```

### 3. Selection Sort

Repeatedly finds the minimum element in the unsorted portion and swaps it with the first unsorted element.

- **Approach:** Comparison-based, in-place sorting.
- **Time Complexity:** \(O(n^2)\)
- **Space Complexity:** \(O(1)\)
- **Order:** Ascending.

```python
def SelectionSort(list):
    listSize = len(list)
    for i in range(0, listSize - 1):
        min = 10000000
        for j in range(i, listSize):
            if list[j] < min:
                min = list[j]
                minIndex = j
        list[minIndex], list[i] = list[i], list[minIndex]
    return list
```

### 4. Insertion Sort

Builds the sorted portion one element at a time by inserting each element into its correct position.

- **Approach:** Comparison-based, in-place sorting.
- **Time Complexity:** \(O(n^2)\)
- **Space Complexity:** \(O(1)\)
- **Order:** Ascending.

```python
def InsertionSort(list):
    listSize = len(list)
    for i in range(listSize):
        for j in range(i, 0, -1):
            if list[j] < list[j - 1]:
                list[j], list[j - 1] = list[j - 1], list[j]
    return list
```
